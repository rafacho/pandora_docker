```python
#!/usr/bin/env python

import numpy as np
from PIL import Image

# arr = np.random.randint(0,256, 100*100) #example of a 1-D array
sz = 129
arr = np.array([0,0], dtype='d')
arr.resize((sz,sz))
# print(arr.dtype)
# test = np.array([0], dtype='d')
# test.resize(sz)
# # print(test)
# print(test.dtype) 

hmax = 0.1 
offset = 1 #hmax
omega = 1; # [Hz]
theta = 0


for j in range(sz):
    for i in range(sz):        
        seno =  np.sin(omega * i  + theta) + offset
        # test[i] = seno
        # print(seno)
        # print(angle)
        arr[i,j] = seno * (255/2)    


im = Image.fromarray(arr, mode="L")
im.show()
im.save("image.png", format="PNG")
print(im.size)

```
